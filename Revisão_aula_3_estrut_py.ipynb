```python
# Atividade 1 - Referências

# Cria uma lista e atribui à variável 'a'
a = [10, 20, 30]

# 'b' recebe a mesma referência de 'a'
# As duas variáveis apontam para o mesmo objeto
b = a

# Altera a lista através de 'b'
b.append(40)


# 1 - Qual será a saída?
print("--- 1. Qual será a saída? ---")

print(f"Valor de a: {a}")
print(f"Valor de b: {b}")


# 2 - 'a' e 'b' representam o mesmo objeto?
print("\n--- 2. 'a' e 'b' representam o mesmo objeto? ---")

mesmo_objeto = id(a) == id(b)

print(f"id(a): {id(a)}")
print(f"id(b): {id(b)}")
print(f"Mesmo objeto? {mesmo_objeto}")


# 3 - Por quê?
print("\n--- 3. Por quê? ---")

print("Porque 'b = a' não cria uma nova lista.")
print("'a' e 'b' apontam para o mesmo objeto na memória.")


# 4 - O que aconteceria com b = a.copy()?
print("\n--- 4. Utilizando a.copy() ---")

# Cria uma cópia independente de 'a'
copia = a.copy()

print(f"Valor de a: {a} | id(a): {id(a)}")
print(f"Valor da cópia: {copia} | id(copia): {id(copia)}")
print(f"a e copia apontam para o mesmo objeto? {id(a) == id(copia)}")

# Altera somente a cópia
copia.append(50)

print("\nApós copia.append(50):")
print(f"a: {a}")
print(f"copia: {copia}")
```

```python
# EXERCÍCIO 2 - CONSTRUINDO UMA CADEIA

# Cria a classe Node, que representa cada nó da cadeia.
class Node:

    # Método construtor: é executado quando um novo nó é criado.
    def __init__(self, valor):

        # Armazena o valor dentro do nó.
        self.valor = valor

        # Inicialmente, o nó não aponta para nenhum outro nó.
        # Por isso, 'proximo' começa com None.
        self.proximo = None


# Cria o primeiro nó e armazena o valor "A".
node1 = Node("A")

# Cria o segundo nó e armazena o valor "B".
node2 = Node("B")

# Cria o terceiro nó e armazena o valor "C".
node3 = Node("C")


# Liga o primeiro nó ao segundo.
# O atributo 'proximo' de node1 passa a apontar para node2.
node1.proximo = node2

# Liga o segundo nó ao terceiro.
# O atributo 'proximo' de node2 passa a apontar para node3.
node2.proximo = node3

# O node3 continua com proximo = None,
# pois ele é o último nó da cadeia.


# Cria uma variável auxiliar chamada 'atual'.
# Ela começa apontando para o primeiro nó da cadeia.
atual = node1


# Enquanto 'atual' não for None,
# o laço continuará sendo executado.
while atual is not None:

    # Exibe o valor armazenado no nó atual.
    print(atual.valor)

    # Faz 'atual' avançar para o próximo nó da cadeia.
    atual = atual.proximo
```

```python
# EXERCÍCIO 3 - DEPURAÇÃO DO PERCURSO

# Cria a classe Node, que representa cada nó da cadeia.
class Node:

    # Método construtor: é executado quando um novo nó é criado.
    def __init__(self, valor):

        # Armazena o valor dentro do nó.
        self.valor = valor

        # Inicialmente, o nó não aponta para nenhum outro nó.
        self.proximo = None


# Cria três nós.
node1 = Node("10")
node2 = Node("20")
node3 = Node("30")

# Liga os nós formando a cadeia.
node1.proximo = node2
node2.proximo = node3

# A variável 'atual' começa apontando para o primeiro nó.
atual = node1


print("--- 1. Qual variável está incorreta? ---")
print("A variável é 'atual', pois ela precisa ser atualizada durante o percurso.")


print("\n--- 2. Qual é o valor dela? ---")
print("O valor armazenado no nó atual é:", atual.valor)


print("\n--- 3. Qual objeto ela referencia? ---")
print("A variável 'atual' referencia inicialmente o objeto node1.")
print("atual é node1?", atual is node1)


print("\n--- 4. Essa referência deveria mudar? ---")
print("Sim. 'atual' deve avançar para o próximo nó da cadeia.")


print("\n--- 5. Qual condição encerra o while? ---")
print("O while termina quando 'atual' recebe None.")


print("\n--- Percorrendo a cadeia corretamente ---")

# Percorre a cadeia enquanto 'atual' não for None.
while atual is not None:

    # Exibe o valor do nó atual.
    print(atual.valor)

    # Atualiza a referência para o próximo nó.
    # Sem esta linha, o programa entraria em loop infinito.
    atual = atual.proximo #Serve para avançar o ponteiro
```
