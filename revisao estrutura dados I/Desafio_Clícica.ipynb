```python
# DESAFIO - ATENDIMENTO PRIORITÁRIO

# Classe que representa os dados de cada paciente.
class Paciente:
    def __init__(self, nome: str, idade: int):

        # Nome do paciente.
        self.nome = nome

        # Idade do paciente.
        self.idade = idade

        # Define se o paciente possui prioridade.
        # Neste exercício, pacientes com 60 anos ou mais
        # são considerados prioritários.
        self.eh_prioritario = idade >= 60

    # Define como o paciente será exibido na tela.
    def __repr__(self):

        if self.eh_prioritario:
            status = "PRIORITÁRIO"
        else:
            status = "NORMAL"

        return f"[{self.nome}, {self.idade} anos - {status}]"

# Classe que representa cada nó da fila.
class Node:

    def __init__(self, paciente: Paciente):

        # Armazena o paciente dentro do nó.
        self.dado = paciente

        # Referência para o próximo nó da fila.
        # Inicialmente não existe próximo nó.
        self.proximo = None

# Classe responsável por controlar a fila de atendimento.
class FilaClinica:

    def __init__(self):

        # Referência para o primeiro nó da fila.
        self.inicio = None

        # Referência para o último nó da fila.
        self.fim = None

        # Quantidade de pacientes aguardando.
        self.tamanho = 0

    # Verifica se a fila está vazia.
    def esta_vazia(self):

        return self.inicio is None

    # Adiciona um novo paciente à fila.
    def adicionar(self, nome: str, idade: int):

        # Cria um novo paciente.
        novo_paciente = Paciente(nome, idade)

        # Cria um novo nó contendo o paciente.
        novo_no = Node(novo_paciente)

        # CASO 1:
        # Se a fila estiver vazia, o novo nó
        # será o início e também o fim da fila.
        if self.esta_vazia():

            self.inicio = novo_no
            self.fim = novo_no

            self.tamanho += 1

            print(
                f"Paciente {nome} adicionado "
                "como o primeiro da fila."
            )
            return

        # CASO 2:
        # Se o paciente for prioritário.
        if novo_paciente.eh_prioritario:

            # Se o primeiro paciente da fila for normal,
            # o novo prioritário entra no início.
            if not self.inicio.dado.eh_prioritario:

                # O novo nó aponta para o antigo início.
                novo_no.proximo = self.inicio

                # O novo nó passa a ser o início da fila.
                self.inicio = novo_no

            else:

                # Começa a percorrer a fila pelo início.
                atual = self.inicio

                # Avança enquanto o próximo paciente
                # também for prioritário.
                while (
                    atual.proximo is not None
                    and atual.proximo.dado.eh_prioritario
                ):
                    atual = atual.proximo

                # O novo nó aponta para o paciente
                # que vinha depois do último prioritário.
                novo_no.proximo = atual.proximo

                # O último prioritário passa a apontar
                # para o novo paciente.
                atual.proximo = novo_no

                # Se o novo nó foi inserido no final,
                # atualiza também a referência fim.
                if novo_no.proximo is None:

                    self.fim = novo_no

        # CASO 3:
        # Se o paciente for normal,
        # ele sempre entra no final da fila.
        else:

            self.fim.proximo = novo_no
            self.fim = novo_no

        # Aumenta a quantidade de pacientes.
        self.tamanho += 1

        print(
            f"Paciente {nome} ({idade} anos) "
            "adicionado à fila."
        )

    # Atende o primeiro paciente da fila.
    def atender(self):

        # Verifica se existe alguém aguardando.
        if self.esta_vazia():

            print("Nenhum paciente aguardando atendimento.")

            return None

        # Guarda o paciente do início da fila.
        paciente_atendido = self.inicio.dado

        # O próximo nó passa a ser o novo início.
        self.inicio = self.inicio.proximo

        # Diminui a quantidade de pacientes.
        self.tamanho -= 1

        # Se a fila ficou vazia,
        # o fim também deve ser None.
        if self.inicio is None:

            self.fim = None

        print(
            f"Atendendo: {paciente_atendido.nome}"
        )
        return paciente_atendido

    # Exibe os pacientes que ainda estão aguardando.
    def listar_espera(self):

        print("\n--- FILA DE ESPERA ---")

        # Verifica se a fila está vazia.
        if self.esta_vazia():

            print("Nenhum paciente aguardando.")

            return

        # Começa pelo primeiro nó.
        atual = self.inicio

        posicao = 1

        # Percorre todos os nós até chegar em None.
        while atual is not None:

            print(f"{posicao}º -> {atual.dado}")

            # Avança para o próximo nó.
            atual = atual.proximo

            posicao += 1

        print(f"Total de pacientes: {self.tamanho}")

# -------------------------------------------------
# TESTE DO PROGRAMA
# -------------------------------------------------

# Cria uma fila de atendimento.
clinica = FilaClinica()

# Adiciona pacientes normais.
clinica.adicionar("João", 30)
clinica.adicionar("Maria", 25)

# Adiciona um paciente prioritário.
clinica.adicionar("Ana", 72)

# Adiciona outro paciente normal.
clinica.adicionar("Pedro", 40)

# Adiciona outro paciente prioritário.
clinica.adicionar("Bento", 80)

# Mostra a fila antes dos atendimentos.
clinica.listar_espera()

print("\n--- INÍCIO DOS ATENDIMENTOS ---")

# Atende o primeiro paciente.
clinica.atender()

# Mostra como ficou a fila.
clinica.listar_espera()
```

```python
# DESAFIO - FILA DE ATENDIMENTO HOSPITALAR PRIORITÁRIO

# Classe que representa cada paciente.
class Paciente:

    def __init__(self, nome: str, idade: int):
        self.nome = nome
        self.idade = idade

        # Pacientes com 60 anos ou mais são prioritários.
        self.eh_prioritario = idade >= 60

    def __repr__(self):

        if self.eh_prioritario:
            status = "PRIORITÁRIO"
        else:
            status = "NORMAL"

        return f"{self.nome} - {self.idade} anos ({status})"

# Cada nó armazena um paciente
# e uma referência para o próximo nó.
class Node:

    def __init__(self, paciente: Paciente):
        self.dado = paciente
        self.proximo = None

# Classe responsável pela fila de atendimento.
class FilaHospital:

    def __init__(self):

        # Primeiro paciente da fila.
        self.inicio = None

        # Último paciente da fila.
        self.fim = None

        # Quantidade de pacientes aguardando.
        self.total_pacientes = 0

    # Verifica se a fila está vazia.
    def esta_vazia(self):

        return self.inicio is None

    # Adiciona um paciente à fila.
    def adicionar(self, nome: str, idade: int):

        novo_paciente = Paciente(nome, idade)

        novo_no = Node(novo_paciente)

        # Caso a fila esteja vazia.
        if self.esta_vazia():

            self.inicio = novo_no
            self.fim = novo_no

            self.total_pacientes += 1

            print(f"Paciente '{nome}' adicionado à fila.")

            return

        # Caso o paciente seja prioritário.
        if novo_paciente.eh_prioritario:

            # Se o primeiro paciente for normal,
            # o prioritário entra no início.
            if not self.inicio.dado.eh_prioritario:

                novo_no.proximo = self.inicio

                self.inicio = novo_no

            else:

                # Percorre os pacientes prioritários.
                atual = self.inicio

                while (atual.proximo is not None and atual.proximo.dado.eh_prioritario):
                    atual = atual.proximo

                # Insere depois do último prioritário.
                novo_no.proximo = atual.proximo

                atual.proximo = novo_no

                # Atualiza o fim caso necessário.
                if novo_no.proximo is None:

                    self.fim = novo_no

        # Pacientes normais entram no final.
        else:
            self.fim.proximo = novo_no

            self.fim = novo_no

        self.total_pacientes += 1

        print(f"Paciente '{nome}' adicionado à fila.")

    # Atende o primeiro paciente.
    def atender(self):

        if self.esta_vazia():

            print("Nenhum paciente aguardando.")

            return None

        paciente_atendido = self.inicio.dado

        # O próximo nó passa a ser o novo início.
        self.inicio = self.inicio.proximo

        # Atualiza o fim caso a fila fique vazia.
        if self.inicio is None:

            self.fim = None

        self.total_pacientes -= 1

        print(f"Atendendo: {paciente_atendido}")

        return paciente_atendido

    # Exibe toda a fila.
    def listar_espera(self):

        print("\n--- FILA DE ESPERA ---")

        if self.esta_vazia():

            print("Fila vazia.")

            return

        atual = self.inicio

        posicao = 1

        while atual is not None:

            print(
                f"{posicao}. {atual.dado}"
            )

            atual = atual.proximo

            posicao += 1

        print(
            f"Total de pacientes: {self.total_pacientes}"
        )

# -----------------------------
# TESTES
# -----------------------------

clinica = FilaHospital()

print("\n--- CHEGADA DOS PACIENTES ---")

clinica.adicionar("João", 30)

clinica.adicionar("Maria", 25)

clinica.adicionar("Ana", 72)

clinica.adicionar("Pedro", 40)

clinica.adicionar("Bento", 80)

clinica.listar_espera()

print("\n--- ATENDIMENTOS ---")

clinica.atender()

clinica.atender()

clinica.listar_espera()

print("\n--- NOVO PACIENTE PRIORITÁRIO ---")

clinica.adicionar("Carlos", 65)

clinica.listar_espera()

print("\n--- ATENDENDO TODOS ---")

while not clinica.esta_vazia():

    clinica.atender()

print("\n--- SITUAÇÃO FINAL ---")

clinica.listar_espera()
```
