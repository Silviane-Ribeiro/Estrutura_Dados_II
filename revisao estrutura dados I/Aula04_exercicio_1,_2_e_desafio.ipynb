```python
# EXERCÍCIO 1 - SIMULADOR "DESFAZER"

# Classe que representa um editor de texto.
class EditorDeTexto:

# O __init__ é executado quando criamos um objeto da classe.
    def __init__(self):

# Cria uma lista vazia que funcionará como PILHA.
# Ela armazenará o histórico das ações realizadas.
        self.historico = []

# Método responsável por executar e armazenar uma nova ação.
    def executar_acao(self, acao: str):

# append() adiciona a nova ação no TOPO da pilha.
        self.historico.append(acao)

        print(f"Ação executada: '{acao}'")

# Método responsável por desfazer a última ação.
    def desfazer(self):

# Verifica se a pilha está vazia.
        if not self.historico:
            print("Nenhuma ação para desfazer!")

# Encerra o método sem retornar uma ação.
            return None

# pop() remove e retorna o elemento do TOPO.
# Portanto, remove a última ação executada.
        ultima_acao = self.historico.pop()
        print(f"↺ Desfazendo: '{ultima_acao}'")

# Retorna a ação que foi removida.
        return ultima_acao

# Método responsável por mostrar o conteúdo da pilha.
    def exibir_estado(self):
        print(f"Estado atual da pilha: {self.historico}")

# Cria um objeto da classe EditorDeTexto.
editor = EditorDeTexto()

# Simula ações realizadas pelo usuário.
editor.executar_acao("Digitar 'Olá'")
editor.executar_acao("Digitar ' Mundo'")
editor.executar_acao("Apagar 'Mundo'")

# Exibe a pilha.
editor.exibir_estado()
print("-" * 30)

# Desfaz a última ação.
editor.desfazer()

# Desfaz novamente.
editor.desfazer()

# Exibe novamente a pilha.
editor.exibir_estado()
```

```python
# Classe responsável pela pilha de ações.
class EditorDeTexto:

# Inicializa a pilha vazia.
    def __init__(self):
        self.historico = []

# Adiciona uma ação no topo da pilha.
    def executar_acao(self, acao: str):
        self.historico.append(acao)
        print(f"Ação executada: {acao}")

# Desfaz a última ação realizada.
    def desfazer(self):

# Verifica se a pilha está vazia.
        if len(self.historico) == 0:
            print("Nenhuma ação para desfazer!")
            return None

# Remove a última ação inserida na pilha.
        ultima_acao = self.historico.pop()
        print(f"Ação desfeita: {ultima_acao}")

        return ultima_acao

# Exibe as ações armazenadas na pilha.
    def mostrar_historico(self):

        if len(self.historico) == 0:
            print("Histórico vazio.")

        else:
            print("Histórico de ações:")

            for acao in self.historico:
                print(acao)
# Cria um objeto da classe EditorDeTexto.
editor = EditorDeTexto()

# Executa algumas ações.
editor.executar_acao("Digitar: Bom dia")
editor.executar_acao("Apagar: dia")
editor.executar_acao("Digitar: tarde")
editor.executar_acao("Substituir: tarde por noite")

# Mostra as ações armazenadas.
editor.mostrar_historico()

# Desfaz a última ação.
editor.desfazer()

# Mostra novamente o histórico.
editor.mostrar_historico()
```

```python
# Importa deque da biblioteca collections.
from collections import deque

# Classe que representa o sistema de impressão.
class Impressora:

# Inicializa o objeto.
    def __init__(self):

# Cria uma FILA vazia utilizando deque.
        self.fila_impressao = deque()

# Adiciona um documento à fila.
    def enviar_documento(self, nome_documento: str):

# append() coloca o documento no FINAL da fila.
        self.fila_impressao.append(nome_documento)

        print(
            f"Documento '{nome_documento}' adicionado à fila."
        )

# Imprime o primeiro documento da fila.
    def imprimir_proximo(self):

# Verifica se a fila está vazia.
        if not self.fila_impressao:
            print("Nenhum documento na fila para imprimir.")

            return None

# popleft() remove o elemento do INÍCIO da fila.
        doc = self.fila_impressao.popleft()
        print(f"Imprimindo: '{doc}'")

        return doc

# Mostra os documentos que ainda estão aguardando.
    def exibir_fila(self):

        print(f"Fila atual de impressão: "f"{list(self.fila_impressao)}"
        )

# Cria um objeto Impressora.
impressora = Impressora()

# Adiciona três documentos à fila.
impressora.enviar_documento("Relatorio_Financeiro.pdf")

impressora.enviar_documento("Trabalho_Faculdade.docx")

impressora.enviar_documento("Boleto_Energia.pdf")

# Exibe a fila.
impressora.exibir_fila()

print("-" * 30)

# Imprime o primeiro documento.
impressora.imprimir_proximo()

# Imprime o próximo.
impressora.imprimir_proximo()

# Exibe o que ainda está na fila.
impressora.exibir_fila()
```

```python
from collections import deque

# Classe responsável por controlar a fila de impressão.
class SpoolerImpressora:

    def __init__(self):

        # Cria uma fila vazia usando deque.
        # O deque é adequado para filas porque permite
        # inserir no final e remover do início.
        self.fila_impressao = deque()

    # Método para adicionar um documento à fila.
    def adicionar_documento(self, documento):

        # O append() adiciona o novo documento
        # no FINAL da fila.
        self.fila_impressao.append(documento)

        print(f"-> Documento '{documento}' adicionado à fila.")

    # Método para imprimir o próximo documento.
    def imprimir_proximo(self):

        # Verifica se a fila está vazia.
        if not self.fila_impressao:

            print("-> Nenhum documento na fila para imprimir.")

            return None

        # O popleft() remove e retorna
        # o PRIMEIRO documento da fila.
        #
        # Isso segue a lógica FIFO:
        # First In, First Out
        # Primeiro que entra, primeiro que sai.
        proximo = self.fila_impressao.popleft()

        print(f"-> Imprimindo: '{proximo}'...")

        return proximo

    # Método para exibir os documentos aguardando impressão.
    def exibir(self):

        # Converte o deque para lista somente
        # para facilitar a exibição na tela.
        documentos = list(self.fila_impressao)

        print("\n--- DOCUMENTOS AGUARDANDO ---")

        # Verifica se existem documentos na fila.
        if documentos:

            print(documentos)

        else:

            print("Fila de impressão vazia.")

# -------------------------------------------------
# PROGRAMA PRINCIPAL
# -------------------------------------------------

# Cria um objeto da classe SpoolerImpressora.
spooler = SpoolerImpressora()


# O while True mantém o menu funcionando
# até que o usuário escolha a opção de sair.
while True:

    print("\n=== SPOOLER DE IMPRESSÃO ===")

    print("1 - Adicionar novo documento")
    print("2 - Imprimir o próximo documento")
    print("3 - Exibir fila de impressão")
    print("4 - Sair")

    # Recebe a opção escolhida pelo usuário.
    opcao = input("Escolha a opção desejada: ")

    # -------------------------------------------------
    # OPÇÃO 1 - ADICIONAR DOCUMENTO
    # -------------------------------------------------

    if opcao == "1":

        # Solicita o nome do documento.
        documento = input("Digite o nome do documento: ")

        # Verifica se o usuário digitou algum nome.
        if documento:

            # Adiciona o documento ao final da fila.
            spooler.adicionar_documento(documento)

        else:

            print("Nome do documento não pode ser vazio.")

    # -------------------------------------------------
    # OPÇÃO 2 - IMPRIMIR PRÓXIMO DOCUMENTO
    # -------------------------------------------------

    elif opcao == "2":

        # Remove e imprime o primeiro documento da fila.
        spooler.imprimir_proximo()

    # -------------------------------------------------
    # OPÇÃO 3 - EXIBIR FILA
    # -------------------------------------------------

    elif opcao == "3":

        # Mostra todos os documentos
        # que ainda estão aguardando impressão.
        spooler.exibir()

    # -------------------------------------------------
    # OPÇÃO 4 - SAIR
    # -------------------------------------------------

    elif opcao == "4":

        print("Programa encerrado com sucesso.")

        # O break encerra o while True.
        break

    # -------------------------------------------------
    # OPÇÃO INVÁLIDA
    # -------------------------------------------------

    else:

        print("Opção inválida. Tente novamente.")
```

```python
from collections import deque

# Classe responsável pelo gerenciamento da triagem hospitalar.
class TriagemHospitalar:

    def __init__(self):

        # Cria uma fila vazia utilizando deque.
        # O deque permite inserir e remover elementos
        # tanto no início quanto no final da fila.
        self.fila_atendimento = deque()

    # Método responsável por adicionar um paciente à fila.
    # Por padrão, o paciente não é prioritário.
    def adicionar_paciente(self, nome, prioritario=False):

        # Se o paciente for prioritário,
        # ele será inserido diretamente no INÍCIO da fila.
        if prioritario:

            # appendleft() adiciona o elemento
            # no início da fila.
            self.fila_atendimento.appendleft(f"[PRIORITÁRIO] {nome}")

            print(f"-> Paciente prioritário '{nome}' "
                "inserido no início da fila.")


        # Caso o paciente não seja prioritário,
        # ele será inserido no FINAL da fila.
        else:

            # append() adiciona o elemento
            # no final da fila.
            self.fila_atendimento.append(f"[PADRÃO] {nome}")

            print(f"-> Paciente '{nome}' ""inserido no final da fila.")

    # Método responsável por chamar
    # o próximo paciente para atendimento.
    def chamar_proximo_paciente(self):

        # Verifica se a fila está vazia.
        if not self.fila_atendimento:

            print("-> Nenhum paciente aguardando na fila.")

            return None

        # popleft() remove e retorna
        # o primeiro paciente da fila.
        proximo = self.fila_atendimento.popleft()

        print(f"-> Chamando para atendimento: {proximo}")

        return proximo

    # Método responsável por exibir
    # os pacientes que estão aguardando atendimento.
    def exibir_fila(self):

        # Converte o deque para uma lista
        # apenas para facilitar a exibição.
        pacientes = list(self.fila_atendimento)

        print(f"\nFila atual de atendimento "f"({len(pacientes)} paciente(s)):" )

        # Verifica se a fila está vazia.
        if not pacientes:

            print("Fila vazia.")

            return

        # Percorre todos os pacientes da fila.
        # enumerate() permite mostrar também
        # a posição de cada paciente.
        for posicao, paciente in enumerate(
            pacientes, 1
        ):

            print(f"{posicao}. {paciente}")

# -------------------------------------------------
# PROGRAMA PRINCIPAL
# -------------------------------------------------

# Cria o objeto que representa
# a triagem do hospital.
hospital = TriagemHospitalar()

# O while True mantém o programa funcionando
# até que o usuário escolha a opção de sair.
while True:

    print("\n=== GERENCIAMENTO DE TRIAGEM HOSPITALAR ===")

    print("1 - Cadastrar novo paciente")
    print("2 - Chamar próximo paciente")
    print("3 - Exibir fila de atendimento")
    print("4 - Sair")

    # Recebe a opção escolhida pelo usuário.
    opcao = input("Escolha a opção desejada: ")

    # -------------------------------------------------
    # OPÇÃO 1 - CADASTRAR PACIENTE
    # -------------------------------------------------

    if opcao == "1":

        # Solicita o nome do paciente.
        nome = input("Nome do paciente: ")

        # Verifica se o usuário informou um nome.
        if not nome:

            print("O nome do paciente não pode ser vazio!")

            # Volta para o início do while,
            # exibindo novamente o menu.
            continue


        # Pergunta se o paciente possui prioridade.
        print("\nClassificação do atendimento:")
        print("S - Paciente prioritário")
        print("N - Paciente padrão")

        prioridade_resp = input("O paciente é prioritário? (S/N): ").upper()


        # Verifica se a resposta foi S.
        # Se for S, prioridade recebe True.
        # Caso contrário, recebe False.
        prioridade = prioridade_resp == "S"

        # Adiciona o paciente à fila.
        hospital.adicionar_paciente(nome, prioritario=prioridade)

        # Exibe a fila depois do cadastro.
        hospital.exibir_fila()

    # -------------------------------------------------
    # OPÇÃO 2 - CHAMAR PRÓXIMO PACIENTE
    # -------------------------------------------------

    elif opcao == "2":

        # Chama e remove o primeiro paciente da fila.
        hospital.chamar_proximo_paciente()


        # Exibe como ficou a fila após o atendimento.
        hospital.exibir_fila()

    # -------------------------------------------------
    # OPÇÃO 3 - EXIBIR FILA
    # -------------------------------------------------

    elif opcao == "3":

        # Mostra todos os pacientes aguardando.
        hospital.exibir_fila()

    # -------------------------------------------------
    # OPÇÃO 4 - SAIR
    # -------------------------------------------------

    elif opcao == "4":

        print("Programa encerrado com sucesso.")

        # break encerra o while True.
        break

    # -------------------------------------------------
    # OPÇÃO INVÁLIDA
    # -------------------------------------------------

    else:

        print("Opção inválida! Tente novamente.")
```
